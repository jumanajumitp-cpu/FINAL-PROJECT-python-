### (iii)Creating Derived Columns
In this step,new columns such as **Age Group**,**BMI**,**BMI Category**,and **Lifestyle activity Level** are created to organize the data into clearer categories for analysis.

```python
# Creating Age groups
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 30, 45, 60, 100],
    labels=['Under 18', '18-30', '31-45', '46-60', 'Above 60']
)
```

```python
#Displaying one random row from each Age Group
for group in df["Age_Group"].unique():
    print(f"\nRandom row for {group}:")
    print(df[df["Age_Group"]==group][["Age","Age_Group"]].sample(1))
```


<p>
    <strong>Info:</strong> This creates a new column called 
    <code>Age_Group</code> by grouping individuals into different age ranges. 
    It helps analyze obesity levels across different age groups and understand 
    how age may be associated with obesity.
</p>

```python
# Creating BMI
df['BMI'] = df['Weight'] / (df['Height'] ** 2)
```

```python
#Displaying one random row from each Age Group
for group in df["BMI"].unique():
    print(f"\nRandom row for {group}:")
    print(df[df["BMI"]==group][["BMI","BMI"]].sample(1))
```

<p>
    <strong>Info:</strong> This calculates Body Mass Index (BMI) using 
    weight in kilograms and height in meters. It creates a new column called 
    <code>BMI</code> to help explore the relationship between BMI and obesity 
    levels.
</p>

```python
# Creating BMI Category
df['BMI_Category'] = pd.cut(
    df['BMI'],
    bins=[0, 18.5, 25, 30, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)
```

```python
#Displaying one random row from each BMI Category
for group in df["BMI_Category"].unique():
    print(f"\nRandom row for {group}:")
    print(df[df["BMI_Category"]==group][["BMI","BMI_Category"]].sample(1))
```

<p>
    <strong>Info:</strong> This code categorizes BMI values into four groups
    using commonly used adult BMI ranges. These categories help explore the
    relationship between BMI and obesity levels. However, caution is needed
    when interpreting these categories for adolescents, as BMI interpretation
    may vary with age and sex. Also checked random row to verify its correct or not.
</p>

```python
# Creating Lifestyle Activity Level
df['Lifestyle_Activity_Level'] = pd.cut(
    df['FAF'],
    bins=[-1, 0, 1, 2, float('inf')],
    labels=['No Activity', 'Low Activity', 'Moderate Activity', 'High Activity']
)
```

<p>
    <strong>Info:</strong> This code groups physical activity frequency
    (<code>FAF</code>) into different activity levels. These custom analytical

    understand how activity frequency may be associated with obesity.

<h4 style="color: #dfe7ed;">Create a Function to Summarize Obesity Levels</h4>

```python
# Function to summarize obesity-level records
def Obesity_Stats(column):
    counts = pd.crosstab(
        df[column],
        df["nobeyesdad"]
    )

    percentages = (
        pd.crosstab(
            df[column],
            df["nobeyesdad"],
            normalize="index"
        ) * 100
    ).round(2)

    return counts, percentages
```

**info** :- Groups records by a selected column and calculates the count and percentage of each obesity level within each category.

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
# Obesity Distribution by Gender
counts, percentages = Obesity_Stats("gender")

print("Obesity Counts by Gender:")
display(counts)

print("Obesity Percentages by Gender:")
display(percentages)
```

**Note** :- To compare obesity-level distributions across gender categories.

```python
# Obesity Distribution by Family History
counts, percentages = Obesity_Stats(
    "family_history_with_overweight"
)

print("Obesity Counts by Family History:")
display(counts)

print("Obesity Percentages by Family History:")
display(percentages)
```

**Note** :- To examine how obesity-level distributions differ according to reported family history of overweight.

```python
# Obesity Distribution by Frequent High-Calorie Food Consumption
counts, percentages = Obesity_Stats("favc")

print("Obesity Counts by FAVC:")
display(counts)

print("Obesity Percentages by FAVC:")
display(percentages)
```

**Note** :- To compare obesity-level distributions between individuals who frequently consume high-calorie food and those who do not.

```python
# Obesity Distribution by Physical Activity
print(
    df.groupby("nobeyesdad", observed=True)
    .agg(
        record_count=("faf", "count"),
        mean_activity=("faf", "mean"),
        median_activity=("faf", "median")
    )
    .round(2)
)
```

**Note** To compare the average and median physical activity frequency across obesity-level categories

#### Multivariate Analysis
It examines how multiple variables interact simultaneously to uncover deeper patterns,combined effects,and factors influencing the target outcome.

```python
# Obesity Level by Gender and Family History
df.groupby(
    ["gender", "family_history_with_overweight", "nobeyesdad"],
    observed=True
).size().reset_index(name="record_count")
```

**Note** :- To compare obesity-level record counts across combinations of gender and family history.

```python
# Obesity Level by Age Group and Gender
df.groupby(
    ["age_group", "gender", "nobeyesdad"],
    observed=True
).size().reset_index(name="record_count")
```

**Note** :- To examine how obesity-level counts vary across age groups and gender.

```python
# Obesity Level by Physical Activity and Eating Habits
df.groupby(
    ["favc", "caec", "nobeyesdad"],
    observed=True
).agg(
    record_count=("weight", "size"),
    mean_weight=("weight", "mean"),
    mean_bmi=("bmi", "mean")
).reset_index().round(2)
df.head()
```

**Note** :- To examine obesity-level records alongside high-calorie food consumption, eating between meals, average weight, and average BMI.
